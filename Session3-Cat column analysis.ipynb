```python
cat=visa_df.select_dtypes(include='object').columns
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10,5))
    colors=['purple','green','blue','brown','red','yellow']
    plt.bar(keys,values,color=colors)
    plt.xlabel("Labels")
    plt.ylabel("count")
    plt.title(f"{i}_bar chart")
    plt.savefig(f"{i}.jpg")
    plt.show()
```

```python

cat=visa_df.select_dtypes(include='object').columns
for i in cat[1:]:
    counts = visa_df[i].value_counts().head(6)
    keys = counts.index
    values = counts.values
    plt.figure(figsize=(10,5))
    colors = ['purple','green','blue','brown','red','yellow']
    plt.pie(values,labels=keys,colors=colors,autopct='%1.1f%%')
    plt.title(f"{i}_pie chart")
    plt.axis('equal')  # makes circle perfect
    plt.savefig(f"{i}.jpg")
plt.show()
```
