**One Hot Encoder**
- One hot means at a time only one value **ON**
  
- Other will be ***OFF**

- ON means one value become **True**

- OFF means one values beacome **False**

- True means one value become **1**

- False means one value become **0**

- first we need to know how many unique labels are there

- for ex:-**case_status** has two unique labels

     - Certified
     - Denied
- so one hot encoder creates two new columns

     - case_status_certified
     - case_status_denied
